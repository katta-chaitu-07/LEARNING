# Eligen values and Eligen vectors

```python
# Suppose you stretch a rubber sheet.

# Most arrows drawn on it will change both:

# Length
# Direction

# But a few special arrows only change their length, not their direction.

# Those special arrows are called Eigenvectors.

# The amount they stretch is called the Eigenvalue.

# Mathematical Definition

# If

# Ax=λx

# where

# A → matrix
# x → eigenvector
# λ → eigenvalue

# then x is an eigenvector of A.?
```

```python
import numpy as np

A = np.array([
    [4,2],
    [1,3]
])

A
```

```python
# working on the eligen values and vectors

values,vectors = np.linalg.eig(A)

print(values)

print(vectors)
```

```python
A = np.array([
    [4,2],
    [1,3]
])

values, vectors = np.linalg.eig(A)

v = vectors[:,0]

print(A @ v)
print(values[0] * v)
```
