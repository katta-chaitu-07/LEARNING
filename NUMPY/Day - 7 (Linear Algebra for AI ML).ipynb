## 4. Matrix Multiplication


```python
# matrix A(2,3) 

A = np.array([
    [1,2,3],
    [4,5,6]
])

# matrix B(3,2)
B = np.array([
    [7,8],
    [9,10],
    [11,12]
])

print(A)

print()
print(B)
```

```python
# one way

A @ B  # matrix multiplication
```

```python
# anoter way 

np.matmul(A,B)


# (m × n)

# ×

# (n × p)

# ↓

# (m × p)
```

## 5. Transpose

## 7. Determinant


```python
# The determinant tells us important properties about a matrix, such as whether it is invertible.

# For a 2×2 matrix:

# A=[
# [a,c],
# [b,d]
# ]

# The determinant is:

# ad−bc
```

```python
A = np.array([
    [9,2],
    [3,4]
])

A
```

```python
det = np.linalg.det(A)

print(det)
```

```python
# Interpretation
# Determinant ≠ 0 → Matrix has an inverse.
# # Determinant = 0 → Matrix is singular (cannot be inverted).
```

## 8 . inverse

```python
A = np.array([
    [2,1],
    [5,3]
])

A_inv = np.linalg.inv(A)

print(A_inv)
```

## 9. Eigenvalues & Eigenvectors

```python
# This is one of the most important ideas in linear algebra.

# Imagine stretching an object.

# Most directions change.

# Some special directions only get scaled, not rotated.

# Those special directions are eigenvectors.

# The amount they are stretched is the eigenvalue.
```

```python
A = np.array([
    [4,2],
    [1,3]
])

values, vectors = np.linalg.eig(A)

print(values)
print()
print(vectors)
```
