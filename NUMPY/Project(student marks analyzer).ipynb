# Task - 3 student average

```python
# Average of each student

average =  np.mean(data_set,axis=1)

average
```

# Task - 4 subject wise average

```python
sub_avg = np.mean(data_set,axis = 0)

sub_avg
```

# Task - 5 Highest scorer

```python

top = np.sum(data_set,axis = 1)


print("student : ", np.argmax(top) +1)

print("marks : ",np.max(top))
```

# Task - 6 Lowest scorer

```python
low = np.sum(data_set,axis = 1)

low
```

```python
# finding the low scorer

print("student : ", np.argmin(low))

print("marks : ", np.min(low))
```

# Task - 7 find the top scorer in each subject

```python

sub_top = np.max(data_set,axis = 0)

sub_top
```

# Task - 8 Failed students

```python
# failed students 

rows , cols = np.where(data_set<40)

rows
```

# Task 9  Finding the Grade

```python

# finding the average

for x,avg in enumerate(average):

    if(avg>90):
        print(f"student {x+1} -> A")
    elif(avg > 80):
        print(f"student {x+1} -> B")

    elif(avg > 70):
        print(f"student {x+1} -> C")
    elif(avg > 60):
        print(f"student {x+1} -> D")
    elif(avg < 60):
        print(f"student {x+1} -> F")
```
