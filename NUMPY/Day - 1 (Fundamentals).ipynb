## 3.zeros()

```python
import numpy as np

zero = np.zeros(3) # creates the 1 d array with all elements as zeros , 3 means number of elements

print(zero)
```

```python
#creates the 2d array with all elements as zeros

zero_matrix = np.zeros((3,4))

zero_matrix
```

## 4.ones()

```python
# creates the 1 d array with all elements as one

one = np.ones(3)
# 3 is the number of elements
one
```

```python
# creates the 2 d array with all elements as one

one_matrix = np.ones((4,3))

# here rows = 4 and columns = 3
one_matrix
```

## Attributes

```python
# shape of the array

import numpy as np

arr = np.arange(1,21)

arr = arr.reshape(4,5)

print(arr)

print()

print("The shape of the matrix is : ", arr.shape)
print()
print("rows : 4 ")
print()
print("columns : 5")
```
