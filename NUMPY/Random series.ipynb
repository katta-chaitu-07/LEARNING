## Generating a float value


```python
import numpy as np

float_val = np.random.rand()

print(float_val)
```

## Generating multiple float values



```python
arr = np.random.rand(4)

# here we are generating 4 float vlaues
print(arr)
```

## Generaing a random matrix

```python
matrix = np.random.rand(3,4)

print(matrix)
```

# PART 2 

## Generating integer values

```python
import numpy as np

int_val = np.random.randint(1,10,)

# start is inclusive

# end is exclusive

# size = number of elements 

print(int_val)
```

## Generating multiple integer values


```python
arr = np.random.randint(1,100,size=10)

# we are generating 10 values 
print(arr)
```

```python
# Generating integer matrix

matrix = np.random.randint(1,100,size=(3,4))

print(matrix)
```

# PART 3

## Random seed



```python
import numpy as np
np.random.seed(42)

arr = np.random.randint(1,100,size = (10,5))

print(arr)
```
