# Part 2

## Group by

```python
Instead of manually filtering each department, groupby() does it automatically.
```

```python
# Finding the average salary of each department
```

```python
data.groupby('Department')['Salary'].mean()
```

## max salary from each department

```python
data.groupby('Department')['Salary'].max()
```

## min salary from each department

```python
data.groupby('Department')['Salary'].min()
```

## count employee from each department

```python
data.groupby('Department')['EmployeeID'].count()
```

## Group by multiple columns

```python
data.groupby(
    ['Department','Gender']
)['Salary'].mean()
```

```python
# Example - 1 , display the employee name from each department whose getting max salary 

data.groupby(
    ['Department','Name']
)['Salary'].max()
```

## Part 4 Aggregrate functions

```python
# sum of all employee salaries

data['Salary'].sum()
```

```python
# Max salary from all employee salaries

data['Salary'].max()
```

```python
# Min salary from all employee salaries

data['Salary'].min()
```

```python
# average salary form all employee salaries

data['Salary'].mean()
```

```python
# standard deviation

data['Salary'].std()
```

```python
# variance

data['Salary'].var()
```

# part 4 agg()

```python
# instead of calculating one by one ,we can calculate all together

data.groupby('Department')['Salary'].agg(
    ['mean','max','min','std','var']
)
```

```python
# multiple columns

data.groupby('Department').agg(
    {
        'Salary' : ['mean','max','min','count'],
        'Bonus' : 'sum',
        'Rating' : 'mean'
    }
)
```

# part 6 transform

```python
Unlike aggregation, which returns one value per group, transformation returns one value for every row.

Example:

Every employee receives the average salary of their own department as a new column.
```

```python
data['Dept_avg'] = data.groupby('Department')['Salary'].transform('mean') 


data['Dept_avg']
```

```python
data.head()
```

```python
data.loc[0:5,['Department','Dept_avg']]
```

# part 7 rank()

```python
Ranking helps identify the highest and lowest values within each group.

Examples:

Highest-paid employee in every department
Top performer in every city
Rank employees based on experience

Function
rank()
8. Top Records
```

```python
# Rank employees by salary within each department.

data['Rank'] = data.groupby('Department')['Salary'].rank(ascending =False)


data['Rank']
```

# Part 8 Top Records

```python
8. Top Records

Sometimes we need only the highest or lowest values.

Examples:

Top 5 salaries
Top employee from each department
Top-rated employees
Functions
nlargest()
nsmallest()
```

```python
# highest salary

data.nlargest(1,'Salary')
```

```python
# second latgest salary

data.nlargest(2,'Salary')
```

```python
# lowest slary 

data.nsmallest(1,'Salary')
```

```python
# second lowest salary 

data.nsmallest(2,'Salary')
```

```python
# Top salary from each department

data.groupby('Department').apply(

    lambda x: x.nlargest(1,'Salary')
)
```
