## 6 . Boolean indexing

## single conditon

```python
# this retrives the columns based on the certain condition is satisfied

data['Marks']>90
```

```python
#Example 2

data[data['Marks']>90]
```

```python
# Example 3

data[data['Department'] == 'CSE']
```

```python
# Example 4 display the students whose gender is femal

data[data['Gender'] == 'Female']
```

## multiple conditons

```python
# single condition

# selecting people from city hyderabad
```

```python
data[data['City'] == 'Hyderabad']
```

```python
# selecting male gender people

data[data['Gender'] == 'Male']
```

```python
# both people from city hyderabab and male gender

data[(data['City'] == 'Hyderabad') & (data['Gender'] == 'Male')]
```

```python
# Example 2
```

```python
# retrive the female people and age greater > 18

data[(data['Gender'] == 'Female') & (data['Age'] > 18)]
```

```python
# or condition
```

```python
data[(data['Gender'] == 'Female') | (data['Age'] > 18)]
```

```python
# using not conditon ~

data[~(data['Department'] == 'CSE')]
```
