## 5 . apply()

```python
One of the most important Pandas functions.

It applies a function to each element or row.
```

```python
# finding the len of the names

data['Name'].apply(len)
```

```python
# apply on functions

def major_or_not(age):
    if age>=18:
        return "Major"
    else:
        return "Minor"

data['Age_status'] = data['Age'].apply(major_or_not)    


data['Age_status']
```
