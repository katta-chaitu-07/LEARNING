## Load the dataset

```python
# import statement

import pandas as pd

data = pd.read_csv("Pandas_Day4_Practice_Dataset (1).csv")

data = pd.DataFrame(data)

data
```

## 2 . isnull()

```python
# Checking for missing values

# True means missing values 

# False means value exist

data.isnull()
```

```python
# Count the missing values

data.isnull().sum()

# This tells us exactly how many missing values exists in each column
```

## 3 . notnull()

```python
# opposite to isnull()

# True = Value exist 

# False - missing value exist
```

```python
data.notnull()
```

```python
data.notnull().sum()

# Tells us exactly how many value exist for each column
```

## 4 . dropna()

```python
Removes the missing values from the dataset

Note :- if it finds any missing value in the row , it deletes the entire row 

Note :- use inplace = True , so the changes will effect the original dataset ,
other wise it updates the dataframe(data in our example) does not effect the original dataset
```

```python
## removes the missing values 

data.dropna()

# i didn't use the inplace = True , because of my further examples 
```

```python
# it does not effect my original dataframe
data
```

```python
import pandas as pd

data = pd.read_csv("Pandas_Day4_Practice_Dataset (1).csv")

data
```

## Remove the row if any value contains missing value


```python
data.dropna(how = 'any')
```

```python
data
```

## Remove the row if all values are null

```python
# it removes the row if all values are null in the record other wise no deletion occured

data.dropna(how = 'all')
```

```python
data
```

## 5 . fillna()

```python
# instedaing of deleting the rows , it fills the missing values
```

## filling with 0

```python
data.fillna(0)

# see the missing value are filled with 0,doesnot modify the orifinal dataframe
```

```python
data
```

## -> Fill with mean 

```python
# I am filling the age columns

data['Age']=data['Age'].fillna(int(data['Age'].mean()))
```

```python
data
```

## Fill with median 

```python
# useful when our dataset contain outliers
```

```python
data['Attendance'] = data['Attendance'].fillna(data['Attendance'].median())
```

```python
data
```

## Forward fill 

```python
copies the previous value
```

```python
# in this example i am using the city column


data['City'] = data['City'].ffill()
```

```python
data
```

## Back ward fill

```python
data['City'] = data['City'].bfill()

data
```

# 6 . Replace the value with other value

```python
data['City'].replace('Hyderabad','Hyd')
```

```python
data
```
