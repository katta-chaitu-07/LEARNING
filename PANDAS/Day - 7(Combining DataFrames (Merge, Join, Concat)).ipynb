# part 2 merge()

```python
It is used to combine two DataFrames based on one or more common columns, just like SQL JOIN.


Parameters
Parameter	Description
left	    First DataFrame
right	    Second DataFrame
how	        Type of join
on	        Common column name
left_on	    Column name in left DataFrame
right_on	Column name in right DataFrame
suffixes	Rename duplicate column names    
```

```python
import pandas as pd

employees = pd.DataFrame({
    'EmpID':[101,102,103,104],
    'Name':['Alice','Bob','Charlie','David'],
    'DeptID':[1,2,2,3]
})

employees
```

```python
departments = pd.DataFrame({
    'DeptID':[1,2,4],
    'Department':['HR','IT','Sales']
})

departments
```

## 1 . inner join()

```python
pd.merge(
    employees,
    departments,
    on = 'DeptID'
    
)

# by default inner join

# dept 3 and 4 is removed , because it is not present in both tables
```

## 2 . Left join()

```python
pd.merge(
    employees,
    departments,
    on = 'DeptID',
    how = 'left'
)

# common from the both sides, retrive all values from the left 
```

## 3 .Right join()



```python
pd.merge(
    employees,
    departments,
    on = 'DeptID',
    how = 'right'
)

# common from the both dataframes , and all rows from the right dataframe(departments)
```

## 4 . outer()

```python
#combine the both dataframes , which include all the rows from the both dataframes

pd.merge(
    employees,
    departments,
    on ='DeptID',
    how = 'outer'
)
```

## 5 . merge on different column names

```python
Sometimes the common columns have different names.
```

```python
employees = pd.DataFrame({
    'EmployeeID':[101,102],
    'Dept':[1,2]
})

departments = pd.DataFrame({
    'DepartmentID':[1,2],
    'Department':['HR','IT']
})
```

```python
print(employees)

print(departments)
```

```python
pd.merge(
    employees,
    departments,
    left_on = 'Dept',
    right_on = 'DepartmentID'
)
```

# Merge on Multiple Columns

```python
df1 = pd.DataFrame({
    'ID':[1,1,2],
    'Year':[2024,2025,2025],
    'Marks':[80,85,90]
})

df2 = pd.DataFrame({
    'ID':[1,2],
    'Year':[2025,2025],
    'Grade':['A','A+']
})
```

```python
print(df1)

print()

print(df2)
```

```python
pd.merge(
    df1,
    df2,
    on = ['ID','Year']
)
```
