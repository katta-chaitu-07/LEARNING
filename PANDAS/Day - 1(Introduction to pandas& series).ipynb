## 2. Series Datatype

## 3 . creating a series 

## From a list

```python
import pandas as pd

arr = pd.Series([85,70,82,90,95])

print(arr)

# we see that the values are labeled 
```

## From a Numpy array



```python
import numpy as np

import pandas as pd

arr = np.random.randint(1,100,size = 10)

arr
```

```python
# we convert the numpy array to the series
```

```python
series = pd.Series(arr)

print(series)
```

## From a dictonary

```python
dict = {
    'maths':90,
    'science' : 85,
    'english' : 88,
    'social' : 70
}

# converting the dict to series

series = pd.Series(dict)

print(series)

# Here notice one thing , the key in the dictonary becomes the indexed labels       
```

## creating the custom index

```python
# we can create the custom index 

custom = pd.Series(
    [90,85,88,70],
    index = ['maths','science','english','social']
)

print(custom)
```

## 4 . Accessing the elements

## 5 . slicing

```python
import numpy as np
import pandas as pd

array = np.random.randint(1,100,size =8)

print(array)
```

```python
# converting it into the series dataframe
pd_array = pd.Series(array)

print(pd_array)
```

```python
# slicing

# syntax array(start:end:step)

print(pd_array[1:5:2])

```

## 7 . mathematical operations


```python

import numpy as np
import pandas as pd

num = np.random.randint(1,99 , size = 6)

print(num)
```

```python
# converting it into the series 

number = pd.Series(num)

print(number)
```

```python
# Addtion , adding 5 to all values

print(number+5)
```

```python
# subtraction , subtract 5 from all values

print(number-5)
```

```python
# multiplication , multiply each value with 2

print(number *2)
```

```python
# Division , divide the each value with 2

print(number /2)
```

## 10 . Handling missing values

```python
import pandas as pd
import numpy as np
data = pd.Series([10,5,np.nan,15,np.nan,24,30,np.nan])

data
```

```python
#checking for missing values
```

```python


print(data.isnull())

# it returns a boolean value

# False means , it contains values

# True means , it contain null values (there is no values)

```

```python
# checking for non missing values
```

```python
print(data.notnull())

# it returns a boolean value

# True = it contains values

# False  = it contain null value
```
