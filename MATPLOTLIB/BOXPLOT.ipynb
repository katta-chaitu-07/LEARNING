# BOX PLOT

**A Box Plot (also called a Box-and-Whisker Plot) is used to summarize the distribution of numerical data.

It helps you quickly identify:

✅ Minimum value
✅ Maximum value
✅ Median
✅ Quartiles (Q1 and Q3)
✅ Interquartile Range (IQR)
✅ Outliers

It is one of the most important plots in Data Analysis, Machine Learning, and Statistics.

Why do we use a Box Plot?

Suppose you have marks of students:

marks = [45, 50, 55, 60, 62, 65, 68, 70, 72, 75, 80, 85, 90]

A histogram shows the distribution, but it doesn't clearly show:

Median
Quartiles
Outliers

A box plot summarizes all of these in one graph.**

## BASIC STRUCTURE


```python
import matplotlib.pyplot as plt

plt.figure(figsize = (8,6))

marks = [45,50,55,60,65,62,68,70,72,75,80,85,90]

plt.boxplot(marks)

plt.title("BOX PLOT")

plt.show()
```

# CUSTIMIZATION

```python
import matplotlib.pyplot as plt

plt.figure(figsize = (8,6))

marks = [45,50,55,60,65,62,68,70,72,75,80,85,90]

plt.boxplot(marks,
            color = "orange",
            edgecolor = "black"
           )

plt.title("BOX PLOT")

plt.show()
```
