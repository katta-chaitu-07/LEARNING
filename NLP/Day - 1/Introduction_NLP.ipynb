## 1. What is NLP

```python
NLP = Natural Language Processing

It is a field of AI that deals with enabling computers to process, understand, analyze, and generate human language.

For example:

"I really enjoyed this movie."

A human immediately understands:

Subject → I
Emotion → positive
Object → movie
Opinion → enjoyed

But a computer initially sees something closer to:

I
really
enjoyed
this
movie

The job of NLP is to build methods that allow the computer to extract useful information and meaning from language.
```

## 2. Why is NLP necessary ?


```python
Computers are excellent at structured data:

Age = 21
Salary = 50000
Experience = 2

But human language is unstructured.

For example:

"Hey, I bought this phone yesterday.
Honestly, the camera is amazing but the battery sucks."

What does the computer need to understand?

Product → phone
Camera → positive
Battery → negative
Overall → mixed

That's where NLP comes in.

NLP is used for tasks such as translation, chatbots, text classification, summarization, question answering, and speech-related systems.
```

## 3 . Natural Language vs Programming Language

```python
3. Natural Language vs Programming Language

This is very important.

Natural language

Languages humans use:

English
Telugu
Hindi
French
Japanese
...

Example:

"Can you send me the file?"
Programming language

Languages designed for computers/programming:

Python
C++
Java
JavaScript
...

Example:

print("Send me the file")

Programming languages generally have strict syntax.

Natural language is much more complicated because it contains:

ambiguity
context
slang
sarcasm
spelling mistakes
multiple meanings
incomplete sentences

Example:

"I saw the man with the telescope."

Who has the telescope?

You?

or

the man?

Humans can often resolve this from context. Computers have to model such relationships.
```

## 4. How does a computer understand text?

```python
Computers fundamentally operate on numbers.

So we need to transform:

I love NLP

into some numerical representation.

A very simplified example:

I     → [0.12, 0.45, 0.76]
love  → [0.91, 0.23, 0.44]
NLP   → [0.67, 0.81, 0.32]

These numerical representations allow ML/deep-learning models to work with language.

Later you'll learn different approaches:

Text
 ↓
Tokenization
 ↓
Numerical representation
 ↓
Machine Learning / Neural Network
 ↓
Prediction

And eventually:

Text
 ↓
Tokens
 ↓
Embeddings
 ↓
Transformer
 ↓
LLM
 ↓
Generated text

That is the bridge from basic NLP → LLMs.
```

## 6 . Text Generation


```python
Before using traditional NLP models, we often preprocess text.

Example:

"I LOVE Machine Learning!!!"

We might convert it into:

"i love machine learning"


Common preprocessing operations:

Raw Text
   ↓
Lowercase
   ↓
Remove unnecessary characters
   ↓
Tokenization
   ↓
Remove stopwords
   ↓
Stemming / Lemmatization
```

## 7. Tokenization

```python
Breaking the text into small possible units called tokens


Example:

"I love NLP"

Word tokenization:

["I", "love", "NLP"]

Another example:

"Machine learning is powerful"

becomes:

["Machine", "learning", "is", "powerful"]


    Why tokenization?

Because algorithms need to work with individual pieces of text.
```

## 8. stop words 

```python
Some words occur very frequently and may provide limited useful information for certain traditional NLP tasks.

Examples:

the
is
a
an
of
to
in
and

Example:

"I am learning NLP"

Possible stopword removal:

"learning NLP"

But ⚠️ don't memorize:

"Stopwords are always useless."

That's wrong.

Whether to remove stopwords depends on the task.

For example, in sentiment analysis:

"I am not happy"

The word:

not

can be extremely important.
```

## 9 . stemming

```python
Stemming attempts to reduce related words to a common root.

For example:

playing
played
plays

might be reduced to:

play

But it can produce incorrect words.

Example:

studies → studi

So stemming is fast but relatively crude.
```

## 10 . Lemitization 

```python
Lemmatization tries to find the proper base/dictionary form.

Example:

studies → study
running → run

Conceptually:

STEMMING
    ↓
Fast
Rule-based
Can produce invalid words

LEMMATIZATION
    ↓
Linguistically meaningful
Uses word information

We'll study this more deeply later.
```
