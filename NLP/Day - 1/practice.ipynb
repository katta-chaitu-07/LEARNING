## stop words

stop words are the common words with contains information , but this are not important

```python
# stop words 

from nltk.corpus import stopwords

nltk.download('stopwords')

stop_words = set(stopwords.words('english'))

print(stop_words)
```

## Remove the stop words from the text 


```python
# Remove the stop words form the text 

text = 'I am Learning the Natural Langauage Processing'

token = word_tokenize(text)

corpus = []

for x in token:
    if x not in stop_words:
        corpus.append(x)

print(corpus)        
```

## Stemming

```python
# reduce the word to their root form by applying simple rules 

print(corpus)

# importing the porter stemmer 

from nltk.stem import PorterStemmer

ps = PorterStemmer()

for word in corpus:
   print(word ,'->',ps.stem(word))
```

```python
Notice:

studies → studi

studi isn't a proper English word.

That's one of the limitations of stemming.
```

## Lematization

Lemmatization is the process of converting a word into its meaningful dictionary/base form, called a lemma.

```python
# Download the necessary resources 

import nltk

nltk.download('wordnet')

nltk.download('omw-1.4')
```

```python
import nltk

from nltk.stem import WordNetLemmatizer

lema = WordNetLemmatizer()

print(corpus)

for word in corpus:

    print(word ,'->',lema.lemmatize(word))
```
