## 1 .Lower case conversion 

```python
text = "I love NLP!!."

text = text.lower()

print(text)
```

## 2 . Tokenize

```python
import nltk

from nltk.tokenize import word_tokenize

text = "I love Natural language processing"

text = word_tokenize(text)

print(text)
```

## 3 . sentance tokenization

```python
from nltk.tokenize import sent_tokenize

text = "I love NLP . NLP is very interesing . I want to learn it."

sent = sent_tokenize(text)

print(sent)
```

## 4 . Remove punctuation

```python
import string

from nltk.tokenize import word_tokenize

text = "Hello ! , how are you ?"

text = word_tokenize(text)

print(text)

# remove the punctuation

tokens = [word for word in text
           if word not in string.punctuation
           ]

print(tokens)
```

## 5 . Remove Numbers

```python
import nltk

from nltk.tokenize import word_tokenize

text = "I was born in 2007"

token = word_tokenize(text)

print(token)

# remove the year 

tokens = [
    word for word in token
    if not word.isdigit()
          ]


print(tokens)
```

## 6 . remove white space

```python
text = "Hello   How  are   you "

text = text.split()

print(text)

text = " ".join(text)

print(text)
```

## 7 . Stop word removal

```python
import nltk

from nltk.corpus import stopwords

stop_words = set(stopwords.words('english'))

print(stop_words)

print(len(stop_words))


text = "I Love the Learning nlp now "

text = word_tokenize(text)

print(text)

token = [word for word in text
         if word.lower() not in stop_words
         ]

print(token)
```

## 8 . stemming


```python
import nltk
from nltk.stem import PorterStemmer

words = ["Playing","watching","seeing","rider","helped","studies"]

stemmer = PorterStemmer()

for word in words:
    print(word,'->',stemmer.stem(word))

```

## 9 . Lemmatization

```python
import nltk

from nltk.stem import WordNetLemmatizer

words = ["studies","running","cleaning","rider","eating","better","children"]

lemma = WordNetLemmatizer()

for word in words:
    print(word,"->",lemma.lemmatize(word))

    
```
