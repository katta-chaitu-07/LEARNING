## 1. pos(parts of speach) Tagging 



```python
import nltk

nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('averaged_perceptron_tagger')
nltk.download('averaged_perceptron_tagger_eng')
```

```python
# import the libraries 

import nltk
from nltk.tokenize import word_tokenize

text = "Chaitanya is learning natural language processing"

tokens = word_tokenize(text)

print(tokens)

#pos tagging 

pos = nltk.pos_tag(tokens)

print(pos)


for word,tag in pos:
    print(word,'->',tag)
    
```

## 2 . Named entity Recognization


```python
nltk.download('maxent_ne_chunker')
nltk.download('maxent_ne_chunker_tab')
nltk.download('words')
```

```python
import nltk

from nltk.tokenize import word_tokenize

text = "elon musk founded the spacex in america "

tokens = word_tokenize(text)

pos = nltk.pos_tag(tokens)

tree = nltk.ne_chunk(pos)


print(tree)


for chunk in tree:
    if hasattr(chunk, "label"):
        entity = " ".join(word for word, pos in chunk)
        label = chunk.label()

        print(entity, "→", label)
```
