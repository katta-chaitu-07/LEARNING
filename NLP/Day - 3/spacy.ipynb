## implementing the named entity recognization using spacy


```python
import spacy

nlp = spacy.load('en_core_web_sm')

text = "chaitanya is learning the natural language processing "

doc = nlp(text)

for token in doc:
    print(token.text)

print()
print("pos tagging")

print()
for token in doc:
    print(token.text,token.pos_)    
```

```python
text = "chaitanya is studying AI and Machine Learning at Gitam university in Hyderabab.He joined Oasis info byte in August 2026"

nlp = spacy.load('en_core_web_sm')

doc  = nlp(text)

for enties in doc.ents:
    print(enties.text,"->",enties.label_)
```

```python
from spacy import displacy

displacy.render(doc, style="ent", jupyter=True)
```
