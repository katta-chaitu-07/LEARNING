```python
import spacy

resume = """
Chaitanya is an AI/ML student at GITAM University, Hyderabad.

He has skills in Python, C++, Machine Learning, Deep Learning,
Natural Language Processing and SQL.

He completed an internship at Oasis Info Byte in 2026.
"""
```

```python
nlp = spacy.load('en_core_web_sm')

doc = nlp(resume)

skills = [
    'Python',
    'C++',
    'Machine Learning'
    'Deep Learning',
    'Natural Language Processing',
    'SQL'
]

founded_skill = []

for skill in skills:
    if skill.lower() in resume.lower():
        founded_skill.append(skill)


print(founded_skill)


for enti in doc.ents:

    print(enti.text,"->",enti.label_)
```
