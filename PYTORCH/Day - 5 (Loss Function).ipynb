## Loss function

## MSE (Mean squarred Error)

```python
# import the library

import torch

import torch.nn as nn

actual = torch.tensor([10.,20.])

predicted = torch.tensor([8.,18.])

loss_function = nn.MSELoss()

loss = loss_function(predicted,actual)

print("The loss is : ",loss)
```

## MAE(Mean absolute Error)

```python
# import the library
import torch

import torch.nn as nn

# actual tensor
actual = torch.tensor([10.,25.,30])

# predicted tensor

predicted = torch.tensor([8.,28.,32.])

# loss function
loss_function = nn.L1Loss()

# calculate the loss

loss = loss_function(predicted,actual)

print("The total loss : ",loss)
```
