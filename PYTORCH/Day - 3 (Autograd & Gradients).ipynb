## 1 . Basic Autograd


```python
# import the torch library
import torch

# step -1 : - create the scaler tensor
x = torch.tensor(3.0 , requires_grad = True)

#step -2 :- perform the operations
y = x ** 2

#step -3 :- Backward pass
y.backward()

#step -4 :- calculate the grad
print("The x grad : ",x.grad)

```

## 2 . More than one operation

```python
# import the library 

import torch

# 1 . create the scaler tensor with float value
x = torch.tensor(3.0,requires_grad = True)

# 2 . multiple operation
y = 3*x**2 +2*x +5

# 3. backward pass
y.backward()

# 4 . gradient
print(x.grad)

```

## 3 . Multiple variabels

```python
#import the torch library

import torch 

# first variabel
x = torch.tensor(2.0,requires_grad = True)

#second variable 
y = torch.tensor(3.0,requires_grad = True)

z = x*y

print("z = ", z)

# calculate the grad of x 

z.backward()

# gradient of x 
print("The gradient of x is : ",x.grad)
print("The gradient of y is : ",y.grad)
```
