## 1. What is training

Training means repeatedly adjusting the model's parameters so that its predictions become closer to the actual values.

We have a model:

Input → Model → Prediction

Initially, the model doesn't know the correct relationship.

For example:

Input       Actual
  1           20
  2           40
  3           60
  4           80

Our model starts with random parameters.

For example:

weight = 0.4
bias = 0.8

For input 2:

$$ prediction = weight \times x + bias $$ $$ = 0.4(2)+0.8 $$ $$ = 1.6 $$

But the actual answer is:

40

## 2. What is training loop

Instead of training only once:

Input
 ↓
Model
 ↓
Prediction

we repeatedly do:

Input
 ↓
Model
 ↓
Prediction
 ↓
Loss
 ↓
Improve model
 ↓
Repeat

That's the training loop.

## Forward pass

```python
# import the libraries

import torch
from torch.utils.data import Dataset
from torch.utils.data import DataLoader
from torch import nn as nn


# create the dataset 

class MyDataset(Dataset):

    # initlize the dataset

    def __init__(self):

        #input
        
        self.x = torch.tensor(
            [
                [1.0],
                [2.0],
                [3.0],
                [4.0]
            ]
        )

        #output

        self.y = torch.tensor(
            [
                [20.0],
                [40.0],
                [60.0],
                [80.0]
            ]
        )

    # len 

    def __len__(self):

        return len(self.x)

    # get the sample

    def __getitem__(self,index):

        return self.x[index],self.y[index]


# Dataset object

dataset = MyDataset()

# DataLoader 

loader = DataLoader(
        
        dataset,
        batch_size =2,
        shuffle = False
)


#Model

model = nn.Linear(

    in_features = 1,
    out_features = 1
)

# optimizer

optimizer = torch.optim.SGD(
    model.parameters(),
    lr = 0.01

)

# Forward pass

for x_batch,y_batch in loader:

    #prediction

    predicted = model(x_batch)

    print("Before weight :", model.weight)


    # loss function

    loss_function = nn.MSELoss()
    loss = loss_function(predicted,y_batch)
    print("Loss : ",loss)

    #clear the old gradients

    print(optimizer.zero_grad())

    # backward pass

    loss.backward()

    print("Weight gradient : ",model.weight.grad)
    print("Bias gradient : ",model.bias.grad)


    #update weights 

    optimizer.step()

    print("After  : ", model.weight)

    print("After weight updation : ",model.weight)
    print("After bias updation : ",model.bias)


    break


```

## Epoch

An epoch means:

One complete pass through the entire training dataset.

Let's use our current dataset:

x = [1, 2, 3, 4]
y = [20, 40, 60, 80]

We have 4 training samples.

Our DataLoader has:

batch_size = 2

So the DataLoader gives us:

Batch 1 → [1, 2]
Batch 2 → [3, 4]

When the model processes both batches, it has seen all 4 samples once.

That is 1 epoch.



Multiple epochs

If we train for:

epochs = 5

the model sees the entire dataset 5 times:

Epoch 1 → entire dataset
Epoch 2 → entire dataset
Epoch 3 → entire dataset
Epoch 4 → entire dataset
Epoch 5 → entire dataset

```python
# import the libraries

import torch
from torch.utils.data import Dataset
from torch.utils.data import DataLoader
from torch import nn as nn


# create the dataset 

class MyDataset(Dataset):

    # initlize the dataset

    def __init__(self):

        #input
        
        self.x = torch.tensor(
            [
                [1.0],
                [2.0],
                [3.0],
                [4.0]
            ]
        )

        #output

        self.y = torch.tensor(
            [
                [20.0],
                [40.0],
                [60.0],
                [80.0]
            ]
        )

    # len 

    def __len__(self):

        return len(self.x)

    # get the sample

    def __getitem__(self,index):

        return self.x[index],self.y[index]


# Dataset object

dataset = MyDataset()

# DataLoader 

loader = DataLoader(
        
        dataset,
        batch_size =2,
        shuffle = False
)


#Model

model = nn.Linear(

    in_features = 1,
    out_features = 1
)

# optimizer

optimizer = torch.optim.SGD(
    model.parameters(),
    lr = 0.01

)

#loss function

loss_function = nn.MSELoss()

for epoch in range(100):

    for x_batch,y_batch in loader:

        #Forward loss
        predicted = model(x_batch)

        #calcualte loss

        loss = loss_function(predicted,y_batch)

        #clear gradient
        optimizer.zero_grad()

        #backword loss
        loss.backward()

        #update the weights and bias
        optimizer.step()
    print("Epoch : ",epoch+1, "Loss : ",loss.item())    

print("weight : ",model.weight.item())
print("bis : ",model.bias.item())   

x = torch.tensor([3.0])

predicted = model(x)

print(predicted.item())
```
