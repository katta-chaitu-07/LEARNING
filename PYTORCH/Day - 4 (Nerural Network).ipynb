```python
import torch 

import torch.nn as nn

layer = nn.Linear(3,2)

x = torch.tensor([
    [10.,20.,30.],
    [5.0,15.,25.]
])

output = layer(x)

print(output)
```

```python
# check the weights of the network

W = layer.weight

print("Weights : ",W)

# shape of the weight

print(W.shape)


# check the initial bias 

print(layer.bias)

print("shape of the bias : ",layer.bias.shape)
```

## Sequential layer

```python
# import the torch module

import torch

import torch.nn as nn

x = torch.tensor([10.,20.,30.])

model = nn.Sequential(

    nn.Linear(3,4),
    nn.Linear(4,2)
)

# output 

output = model(x)

print(output)
```

## nn.Module

```python
import torch 

import torch.nn as nn

class Mymodule(nn.Module):

    def __init__(self):

        super().__init__()

        self.layer1 = nn.Linear(3,5)

        self.relu = nn.ReLU()

        self.layer2 = nn.Linear(5,3)


    def forward(self,x):

        x = self.layer1(x)

        x = self.relu(x)

        x = self.layer2(x)

        return x



model = Mymodule()


x = torch.tensor([[1.0,2.0,3.0]])


output = model(x)



print("Model is : ")
print(model)


print("The input is : ", x)


print("The output is : ", output)
```
