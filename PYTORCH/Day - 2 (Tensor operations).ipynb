# Arithmetic operations

## 1. Tensor Addition

```python
# import the pytorch library

import torch as t

A = t.tensor([1,2,3,4,5])

B = t.tensor([1,3,2,5,6])

print("Tensor Addition : ", A+B)
```

## 2 . Tensor Subtaction

```python
import torch as t

A = t.tensor([3,2,1])

B = t.tensor([1,3,4])

print("Tenosr Subraction : ",A-B)
```

## 3 . Tensor Multiplication

```python
import torch as t

A = t.tensor([1,2,3])

B = t.tensor([3,2,3])

print("Tensor Multiplicaton : ",A*B)


# Note : It follows the element wise muliplication not matrix multiplication
```

## 4 . Tensor Division

```python
import torch as t

A = t.tensor([2,4,6])

B = t.tensor([1,2,2])

print("Tenosr Division : ",A/B)
```

# Scaler Operations

## add a constant 

```python
# we can perform operations between a tensor and the scalar value (single ) 

import torch as t

x = t.tensor([1,2,3])

print(x+10)
```

## Multiply with constant

```python
import torch as t

x = t.tensor([1,2,3])

print(x*2)
```

## Divide with constant

```python
print(x/2)
```

## Important functions

## sum

```python
x = t.tensor([1,2,3,4,5])

print(t.sum(x))
```

## mean

```python
print(t.mean(x.float()))
```

## maximum

```python
print(t.max(x))
```

## minimum

```python
print(t.min(x))
```

## Matrix Multiplication 

```python
import torch

a = torch.tensor([
    [1,2,3],
    [4,5,6]
])

b = torch.tensor([
    [2,1],
    [7,5],
    [1,4]
])

print("Tensor Multiplication : ")

print(torch.matmul(a,b))
```
